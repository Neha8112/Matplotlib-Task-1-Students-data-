```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Student12csv.csv")

print(df.columns)

c = ['y', 'r', 'b']

Average = df[["S1", "S2", "S3"]].mean()

name = Average.index

plt.title('Sub Wise Avg')

plt.pie(Average,labels=name, colors=c, shadow=True,
explode = (0.05, 0.05, 0.05), autopct = '%1.1f%%')

plt.axis('equal')
plt.legend()
plt.show()
```

# 3. Movie Ratings Visualization

```python
import matplotlib.pyplot as plt

import numpy as np

movies = ['Vivah', 'Spider', 'Avengers', 'Titanic', 'Joker']

imdb = [8.8, 8.6, 8.0, 7.8, 8.4]
rotten_tomatoes = [87, 72, 94, 88, 68]
metacritic = [74, 81, 69, 75, 59]

plt.title("Ratings")
plt.xlabel("Movie")
plt.ylabel("Ratings")

x=np.arange(len(movies))
width=0.25

plt.bar(x - width, imdb, width, label="IMDB")
plt.bar(x, rotten_tomatoes, width, label="Rotten_Romatoes")
plt.bar(x + width, metacritic, width, label="metacritic")

plt.xticks(x, movies, rotation=30)

plt.legend()

plt.grid(axis='y', linestyle='--')

plt.show()
```

# 4. Boxplot of Exam Scores

```python
import matplotlib.pyplot as plt
import pandas as pd
data = {
    "Maths":    [85, 78, 92, 88, 76, 95, 89, 91, 84, 77],
    "Science":  [80, 72, 85, 90, 88, 70, 95, 92, 86, 75],
    "English":  [75, 85, 78, 82, 89, 92, 73, 88, 80, 77]
}
df = pd.DataFrame(data)
plt.boxplot([df["Maths"], df["Science"], df["English"]],
            labels=["Maths", "Science", "English"],
            patch_artist=True,
            boxprops=dict(facecolor="lightblue", color="blue"),
            medianprops=dict(color="red", linewidth=2))
plt.title("Distribution of Exam Scores by Subject", fontsize=14)
plt.ylabel("Scores")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
